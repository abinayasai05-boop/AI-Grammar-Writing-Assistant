```python
import re
import language_tool_python
from textblob import TextBlob
import gradio as gr
```

```python
tool = language_tool_python.LanguageTool("en-US")

print("AI Grammar & Writing Assistant is ready!")
```

```python
def correct_grammar(text):
    if not text or not text.strip():
        return "Please enter some text."

    matches = tool.check(text)

    corrected = language_tool_python.utils.correct(text, matches)

    return corrected
```

```python
def analyze_writing(text):
    if not text or not text.strip():
        return "Please enter some text."

    words = re.findall(r'\b\w+\b', text)
    sentences = re.split(r'[.!?]+', text)
    sentences = [s.strip() for s in sentences if s.strip()]

    paragraphs = [p.strip() for p in text.split("\n") if p.strip()]

    word_count = len(words)
    character_count = len(text)
    sentence_count = len(sentences)
    paragraph_count = len(paragraphs)

    if sentence_count > 0:
        avg_sentence_length = word_count / sentence_count
    else:
        avg_sentence_length = 0

    spelling_errors = len(TextBlob(text).words) - len(words)

    analysis = f"""
Writing Analysis
================

Words              : {word_count}
Characters         : {character_count}
Sentences          : {sentence_count}
Paragraphs         : {paragraph_count}
Average sentence   : {avg_sentence_length:.2f} words

Grammar issues     : {len(tool.check(text))}

Writing Level
-------------
"""

    if avg_sentence_length <= 12:
        analysis += "Easy to read / Simple writing"
    elif avg_sentence_length <= 20:
        analysis += "Intermediate writing"
    else:
        analysis += "Advanced / Complex writing"

    return analysis
```

```python
def improve_writing(text):
    if not text or not text.strip():
        return "Please enter some text."

    corrected = language_tool_python.utils.correct(
        text,
        tool.check(text)
    )

    # Basic writing improvements
    corrected = re.sub(r'\bvery good\b', 'excellent', corrected, flags=re.I)
    corrected = re.sub(r'\bvery bad\b', 'poor', corrected, flags=re.I)
    corrected = re.sub(r'\ba lot of\b', 'many', corrected, flags=re.I)
    corrected = re.sub(r'\bin order to\b', 'to', corrected, flags=re.I)
    corrected = re.sub(r'\bdue to the fact that\b', 'because', corrected, flags=re.I)

    return corrected
```

```python
def writing_score(text):
    if not text or not text.strip():
        return "Please enter some text."

    matches = tool.check(text)
    errors = len(matches)

    words = len(re.findall(r'\b\w+\b', text))

    if words == 0:
        return "No text available."

    error_rate = errors / words

    score = max(0, 100 - (error_rate * 100))

    if score >= 90:
        level = "Excellent"
    elif score >= 75:
        level = "Good"
    elif score >= 60:
        level = "Needs Improvement"
    else:
        level = "Needs Major Improvement"

    return f"""
Writing Quality Score
=====================

Score : {score:.1f}/100
Level : {level}

Grammar Issues : {errors}
Word Count     : {words}
"""
```

```python
def spelling_correction(text):
    if not text or not text.strip():
        return "Please enter some text."

    blob = TextBlob(text)

    corrected = blob.correct()

    return str(corrected)
```

```python
def grammar_assistant(text, task):
    if not text or not text.strip():
        return "Please enter some text."

    if task == "Grammar Correction":
        return correct_grammar(text)

    elif task == "Improve Writing":
        return improve_writing(text)

    elif task == "Spelling Correction":
        return spelling_correction(text)

    elif task == "Writing Analysis":
        return analyze_writing(text)

    elif task == "Writing Score":
        return writing_score(text)

    return text
```

```python
def grammar_check(text):

    if not text or not text.strip():
        return "Please enter some text."

    # Check grammar
    matches = tool.check(text)

    # Correct the text
    corrected_text = language_tool_python.utils.correct(text, matches)

    # Create error report
    error_report = []

    for i, match in enumerate(matches, 1):

        original = text[match.offset:match.offset + match.errorLength]

        replacement = (
            match.replacements[0]
            if match.replacements
            else "No suggestion"
        )

        error_report.append(
            f"{i}. ❌ {original} → ✅ {replacement}\n"
            f"   Reason: {match.message}"
        )

    # Statistics
    words = re.findall(r'\b\w+\b', text)
    sentences = re.split(r'[.!?]+', text)
    sentences = [s for s in sentences if s.strip()]

    word_count = len(words)
    sentence_count = len(sentences)

    # Score
    if word_count == 0:
        score = 0
    else:
        error_rate = len(matches) / word_count
        score = max(0, min(100, 100 - (error_rate * 100)))

    if score >= 90:
        level = "Excellent"
    elif score >= 75:
        level = "Good"
    elif score >= 60:
        level = "Needs Improvement"
    else:
        level = "Needs Major Improvement"

    report = "\n".join(error_report)

    if not report:
        report = "✅ No grammar or spelling errors found!"

    result = f"""
ORIGINAL TEXT
-------------
{text}

CORRECTED TEXT
--------------
{corrected_text}

ERRORS & SUGGESTIONS
--------------------
{report}

WRITING ANALYSIS
----------------
Words      : {word_count}
Sentences  : {sentence_count}
Errors     : {len(matches)}

WRITING SCORE
-------------
{score:.1f}/100

LEVEL
-----
{level}
"""

    return result
```

```python
def check_grammar(text):

    if not text or not text.strip():
        return "Please enter some text."

    try:
        # Find grammar mistakes
        matches = tool.check(text)

        # Correct mistakes
        corrected = language_tool_python.utils.correct(text, matches)

        # Statistics
        words = re.findall(r"\b[\w']+\b", text)
        sentences = re.findall(r"[.!?]+", text)

        word_count = len(words)
        sentence_count = len(sentences)
        error_count = len(matches)

        # Score
        if word_count > 0:
            score = max(
                0,
                round(100 - (error_count / word_count * 100), 2)
            )
        else:
            score = 0

        # Level
        if score >= 90:
            level = "Excellent"
        elif score >= 75:
            level = "Good"
        elif score >= 60:
            level = "Needs Improvement"
        else:
            level = "Needs Major Improvement"

        # Error details
        details = ""

        for i, match in enumerate(matches, 1):

            # Different LanguageTool versions use different attributes
            offset = getattr(match, "offset", 0)

            error_length = getattr(
                match,
                "errorLength",
                getattr(match, "error_length", 0)
            )

            wrong = text[offset:offset + error_length]

            # Suggested correction
            replacements = getattr(
                match,
                "replacements",
                []
            )

            if replacements:
                suggestion = replacements[0]
            else:
                suggestion = "No suggestion"

            # Message
            message = getattr(
                match,
                "message",
                "Grammar or spelling issue"
            )

            details += (
                f"{i}. ❌ {wrong} → ✅ {suggestion}\n"
                f"   Reason: {message}\n\n"
            )

        if not details:
            details = "✅ No grammar or spelling mistakes found!"

        result = f"""
╔══════════════════════════════════════════╗
       ✨ AI GRAMMAR & WRITING ASSISTANT
╚══════════════════════════════════════════╝

📝 ORIGINAL TEXT
──────────────────────────────────────────
{text}


✨ CORRECTED TEXT
──────────────────────────────────────────
{corrected}


🔍 ERRORS & SUGGESTIONS
──────────────────────────────────────────
{details}


📊 WRITING ANALYSIS
──────────────────────────────────────────
Words       : {word_count}
Sentences   : {sentence_count}
Errors      : {error_count}


🎯 WRITING SCORE
──────────────────────────────────────────
{score}/100


🏆 WRITING LEVEL
──────────────────────────────────────────
{level}
"""

        return result

    except Exception as e:
        return f"❌ Error: {str(e)}"
```
